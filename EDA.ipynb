```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv('data/sgemm_product.csv')
```

```python
df.head()
```

```python
df.describe()
```

```python
# UCI veri setindeki orijinal kolon isimlerinde genellikle boşluk bulunur.
# Hata almamak için kolon isimlerini tam olarak yazıyoruz.
run_cols = ['Run1 (ms)', 'Run2 (ms)', 'Run3 (ms)', 'Run4 (ms)']
```

```python
# 4 farklı koşunun ortalamasını alarak hedef değişkeni oluştur
# axis=1 parametresi işlemi sütunlar üzerinde (satır bazında) yapmamızı sağlar
df['avarage_runtime'] = df[run_cols].mean(axis=1)
```

```python
# Orijinal Run kolonlarını sil (Çok Önemli!)
# Eğer bunları modelde bırakırsak, model çalışma süresini tahmin etmek için 
# yine çalışma sürelerini kullanır. Buna "Veri Sızıntısı (Data Leakage)" denir.
df = df.drop(columns = run_cols)
```
